Wiechert-Herglotz teorem glasi:

$$\ln\left(\frac{r_{0}}{r}\right) = \frac{1}{\pi}\int_{0}^{\Delta_{1}}\text{Arch}\left(\frac{p}{p_{1}}\right)d\Delta$$

# Rješenje

Preračunajte zadana vremena iz tablice **u sekunde**!

Izračunajte parametar staze vala prema jednadžbi $p_{i}=\frac{dT}{d\Delta}$. 

Možete koristiti numeričku aproksimaciju $p_{i} = \frac{T_{i + 1} - T_{i-1}}{\Delta_{i + 1} - \Delta_{i-1}}$

Izračunajte omjer parametra staze vala i parametra tražene staze vala $(p_{90})$!

Treba izračunati area kosinus hiperbolni, za kojeg vrijedi sljedeća jednadžba:

$$\text{Arch}(x) = \ln\left(x + \sqrt{x^{2} - 1}\right)\quad \text{za} \quad x\ge 1$$

Sada treba izračunati integral

$$I = \int_{0}^{\Delta_{1}}\text{Arch}\left(\frac{p}{p_{1}}\right)d\Delta$$

Koristite trapeznu formulu!
