# [QuakeML](https://quake.ethz.ch/quakeml/)

**Katalog potresa** je organizirana baza podataka o potresima na nekom području te uključuje minimalno informacije o lokaciji hipocentra i vremenu nastanka svakog potresa te njegovoj magnitudi.

[QuakeML](https://quake.ethz.ch/quakeml/) je de facto standard spremanja kataloga potresa. U biblioteci [ObsPy](https://docs.obspy.org/) je definirana implementacija tog standarda koja je hijerarhijski organizirana u klasi tipa [`Catalog`](https://docs.obspy.org/packages/autogen/obspy.core.event.Catalog.html#obspy.core.event.Catalog).

![Struktura objekta Catalog](https://docs.obspy.org/_images/Event.png)

[QuakeML](https://quake.ethz.ch/quakeml/) dokumenti se mogu učitavati pomoću [ObsPy](https://docs.obspy.org/) funkcije [`read_events()`](https://docs.obspy.org/packages/autogen/obspy.core.event.catalog.read_events.html#obspy.core.event.catalog.read_events).

```python
from obspy import read_events

catalog = read_events("./data/2020_events.xml")  # katalog potresa iz za razdoblje 2020. - 2021. godine

print(catalog)
```

Funkcija [`read_events()`](https://docs.obspy.org/packages/autogen/obspy.core.event.catalog.read_events.html#obspy.core.event.catalog.read_events) vraća objekt tipa [`Catalog`](https://docs.obspy.org/packages/autogen/obspy.core.event.Catalog.html#obspy.core.event.Catalog), koji je pak skup objekata tipa [`Event`](https://docs.obspy.org/packages/autogen/obspy.core.event.event.Event.html#obspy.core.event.event.Event). Kao i u slučaju objekta tipa [`Inventory`](https://docs.obspy.org/packages/autogen/obspy.core.inventory.inventory.Inventory.html#obspy.core.inventory.inventory.Inventory), elementi sadržani unutar tog objekta se mogu tretirati kao elementi bilo kojeg Python iterable objekta.

```python
print(type(catalog))
print(type(catalog[0]))
```

```python
event = catalog[0]
print(event)
```

Svaki potres (*event*) može imati definirano više izvora (npr. ako ga neovisno locira više agencija). Svaki izvor s pripadnim informacijama je spremljen u objekt tipa [`Origin`](https://docs.obspy.org/packages/autogen/obspy.core.event.origin.Origin.html#obspy.core.event.origin.Origin).

```python
print(type(event.origins))
print(type(event.origins[0]))
print(event.origins[0])
```

Analogno vrijedi i za magnitude - svaki potres (*event*) može imati više magnituda (recimo više različitih magnituda koje su izračunate - npr. lokalna (Richterova) magnituda i momentna magnituda). Svaka magnituda je spremljena u objekt tipa [`Magnitude`](https://docs.obspy.org/packages/autogen/obspy.core.event.magnitude.Magnitude.html#obspy.core.event.magnitude.Magnitude).

```python
print(type(event.magnitudes))
print(type(event.magnitudes[0]))
print(event.magnitudes[0])
```

# Korisne metode za rad s katalozima

Klasa [`Catalog`](https://docs.obspy.org/packages/autogen/obspy.core.event.Catalog.html#obspy.core.event.Catalog) sadrži korisne metode koje olakšavaju rad s podacima. Ovdje će biti navedene samo neke od njih.

## Filtriranje potresa prema zadanom uvjetu

Metoda [`filter()`](https://docs.obspy.org/packages/autogen/obspy.core.event.Catalog.filter.html#obspy.core.event.Catalog.filter) se koristi kako bi se katalog filtrirao prema zadanim uvjetima. Na primjer, filtrirat ćemo iz kataloga koji smo učitali samo one potrese koji imaju magnitudu veću od 5.2.

```python
largest_magnitude_events = catalog.filter("magnitude >= 5.2")
print(largest_magnitude_events)
```

## Grafički prikaz potresa iz kataloga

Slično kako je bilo definirano u slučaju klase [`Inventory`](https://docs.obspy.org/packages/autogen/obspy.core.inventory.inventory.Inventory.html#obspy.core.inventory.inventory.Inventory), i u slučaju klase [`Catalog`](https://docs.obspy.org/packages/autogen/obspy.core.event.Catalog.html#obspy.core.event.Catalog) postoji metoda [`plot()`](https://docs.obspy.org/packages/autogen/obspy.core.event.Catalog.plot.html#obspy.core.event.Catalog.plot) koja omogućuje grafički prikaz lokacija potresa u katalogu.

```python
catalog = catalog.filter("magnitude >= 3.5")  # radi preglednosti slike

catalog.plot(projection="local");
```
