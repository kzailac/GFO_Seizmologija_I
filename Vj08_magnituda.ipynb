# Zagrebački potres 22. ožujka 2020. godine

Zagrebački potres 22. ožujka 2020. godine dogodio se točno u 05:24:03 UTC. Lokacija epicentra je bila na 45.87°, 15.99°, a dubina hipocentra je bila na 10.0 km. Lokalna magnituda (podatak iz kataloga) je procijenjena na 5.5.


## Zadatak 1

Procijenite lokalnu magnitudu Zagrebačkog potresa. Koristite za procjenu operativnu formulu za procjenu magnitude na području RH:

$$M_{L} = \log{A_{max}} + 2.094\log{\Delta[^{\circ}]} + 2.190,$$

gdje je $A_{max}$ najveća zabilježena amplituda u $\micro\text{m}$, a $\Delta$ je epicentralna udaljenost u stupnjevima.

Koristite seizmogram postaje KALN (u datoteci [`./data/waveform_KALN.mseed`](./data/waveform_KALN.mseed)). Pripadni inventory je dostupan u datoteci [`./data/CR_stations.xml`](./data/CR_stations.xml).

### 1. korak

Učitajte seizmogram i pripadni inventory. Prikažite zapis.

### 2. korak

Uklonite trend, primijenite taper, i uklonite response kako biste dobili seizmograme u fizičkim jedinicama.

## Zadatak 2

Procijenite sada magnitudu korištenjem IASPEI jednadžbe:

$$M_{L} = \log{A_{H,max}}[\text{nm}] + 1.11\log{R[\text{km}]} - 0.00189R[\text{km}] - 2.09,$$

gdje je $A_{H,max}$ najveća zabilježena amplituda na horizontalnim komponentama u nm, a $R$ je hipocentralna udaljenost u km!

## Zadatak 3

Procijenite magnitudu koristeći jednadžbu prema Herak i sur. (1996):

$$M_{L} = \log{A_{max}} + 1.44\log{\Delta[^{\circ}]} + 2.554,$$

gdje je $A_{max}$ najveća zabilježena amplituda u $\micro\text{m}$, a $\Delta$ epicentralna udaljenost u stupnjevima.
