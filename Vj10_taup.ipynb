# TauP paket

[ObsPy TauP paket](https://docs.obspy.org/packages/obspy.taup.html) je krenuo iz Java TauP Toolkita ([Crotwell et al., 1999](https://www.researchgate.net/publication/237546227_The_TauP_Toolkit_Flexible_Seismic_Travel-Time_and_Raypath_Utilities)) koji se koristi za računanje teorijskih vremena putovanja zraka valova u globalnom 1D sferno simetričnom modelu.

```python
from obspy.taup import TauPyModel

model = TauPyModel(model="iasp91")
```

## Vremena putovanja

Glavna metoda u paketu je [`get_travel_times()`](https://docs.obspy.org/packages/autogen/obspy.taup.tau.TauPyModel.get_travel_times.html#obspy.taup.tau.TauPyModel.get_travel_times) koja vraća vremena putovanja zraka za odabrane faze, epicentralne udaljenosti i dubine izvora u inicijaliziranom modelu. Ako se ne odaberu faze, onda se vraćaju vremena putovanja za sve faze.

```python
arrivals = model.get_travel_times(
    source_depth_in_km=55,
    distance_in_degree=67
)

print(arrivals)
```

```python
arrivals = model.get_travel_times(
    source_depth_in_km=100,
    distance_in_degree=45,
    phase_list=["P", "PSPSPS"]
)

print(arrivals)
```

Metoda [`get_travel_times()`](https://docs.obspy.org/packages/autogen/obspy.taup.tau.TauPyModel.get_travel_times.html#obspy.taup.tau.TauPyModel.get_travel_times) vraća iterabilni objekt čiji elementi su pojedini nailasci, a koji su spremljeni kao objekti tipa [`Arrival`](https://docs.obspy.org/packages/autogen/obspy.taup.helper_classes.Arrival.html#obspy.taup.helper_classes.Arrival). 

Pojedinim parametrima se može pristupiti preko tog objekta, kao npr. parametru zrake vala, vremenu putovanja, upadnom i izlaznom kutu.

```python
arr = arrivals[0]

print(f"p = {arr.ray_param},\nt = {arr.time}s,\ni = {arr.incident_angle},\ni_t = {arr.takeoff_angle}")
```

## Grafički prikaz valnih zraka

Putanje valnih zraka se računaju pomoću metode [`get_ray_paths()`](https://docs.obspy.org/packages/autogen/obspy.taup.tau.TauPyModel.get_ray_paths.html#obspy.taup.tau.TauPyModel.get_ray_paths).

Izračunate valne zrake se mogu prikazati grafički pomoću metode [`plot_rays()`](https://docs.obspy.org/packages/autogen/obspy.taup.tau.Arrivals.plot_rays.html#obspy.taup.tau.Arrivals.plot_rays).

```python
arrivals = model.get_ray_paths(
    source_depth_in_km=500, distance_in_degree=130, phase_list=["ttbasic"]
)
ax = arrivals.plot_rays()
```

Umjesto polarnih koordinata mogu se koristiti i Kartezijeve, što se zadaje parametrom `plot_type`.

```python
ax = arrivals.plot_rays(plot_type="cartesian")
```

## Hodokrone

Hodokrone su krivulje ovisnosti vremena putovanja o epicentralnoj udaljenosti. U modelu s dva sloja iznad poluprostora kakav je balkanski model, ovisnosti vremena putovanja pojedinih faza o epicentralnoj udaljenosti (izvod u skripti!) glase:

* u slučaju žarišta u gornjem sloju:

$$t_{g} = \frac{s}{v_{1}} = \frac{\sqrt{\Delta^{2} + h^{2}}}{v_{1}}$$

$$t_{b_{1}} = \frac{\Delta}{v_{2}} + (2h_{1} - h)\frac{\sqrt{v_{2}^{2} - v_{1}^{2}}}{v_{1}v_{2}}$$

$$t_{n_{1}} = \frac{\Delta}{v_{3}} + (2h_{1} - h)\frac{\sqrt{v_{3}^{2} - v_{1}^{2}}}{v_{1}v_{3}} + 2h_{2}\frac{\sqrt{v_{3}^{2} - v_{2}^{2}}}{v_{2}v_{3}}$$

* u slučaju žarišta u donjem sloju:

$$t_{b_{2}} = \frac{h - h_{1}}{v_{2}\cos i} + \frac{\sqrt{[\Delta - (h - h_{1})\tan i]^{2} + h_{1}^{2}}}{v_{1}}$$

$$t_{n_{2}} = \frac{\Delta}{v_{3}} + \frac{(h_{1} + 2h_{2} - h)\sqrt{v_{3}^{2} - v_{2}^{2}}}{v_{2}v_{3}} + \frac{h_{1}}{v_{1}v_{3}}\sqrt{v_{3}^{2} - v_{1}^{2}}$$
