## Gutenberg-Richter relacija

Broj potresa varira s magnitudom, tako da su potresi manjih magnituda puno češći. Relaciju između magnitude i broja pripadnih potresa opisali su Gutenberg i Richter 1940-tih godina. Ta *Gutenberg-Richterova* relacija glasi:

$$\log{N} = a - bM,$$

gdje je $N$ broj potresa magnitude veće od $M$ u danom vremenskom periodu. Relacija je linearna s konstantama $a$ i $b$.

## Omorijev zakon

Većina naknadnih potresa se dogodi brzo nakon glavnog potresa, a njihova učestalost opada kvazi-hiperbolički. To opadanje učestalosti se opisuje relacijom koja se naziva *Omorijevim zakonom*, a koji glasi:

$$n = \frac{c}{(k + t)^{p}},$$

gdje je $n$ učestalost naknadnih potresa u vremenu $t$ nakon glavnog potresa, a $k$, $c$ i $p$ su konstante.
