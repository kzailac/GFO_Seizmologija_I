# Trapezna formula

Određeni integral funkcije $f(x)$ jednak je ukupnoj površini ispod krivulje $y=f(x)$ u intervalu $[a, b]$.

U slučaju trapezne formule, određeni integral se dobiva sumiranjem površina trapezoida koji povezuju točke

$$(x_{i-1},0), (x_{i}, 0), (x_{i-1}, f(x_{i-1})), (x_{i}, f(x_{i}))$$ 

za svaki podinterval $[x_{i-1}, x_{i}]$. Pri tome treba imati na umu da je svaki trapezoid zbroj kvadra i trokuta

$$(x_{i} - x_{i-1})f(x_{i-1}) + \frac{1}{2}(x_{i}-x_{i-1})[f(x_{i}) - f(x_{i-1})] = \frac{1}{2}[f(x_{i}) + f(x_{i-1})](x_{i}-x_{i-1})$$

Za primjer, koristit ćemo jedan trapezoid da aproksimiramo interval:

$$\int_{0}^{1}e^{-x^{2}}dx$$

Najprije ćemo nacrtati krivulju $y = e^{-x^{2}}$ i trapezoid u intervalu $[0, 1]$.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-0.5,1.5,100)
y = np.exp(-x ** 2)
plt.plot(x,y)

x0 = 0; x1 = 1;
y0 = np.exp(-x0 ** 2) 
y1 = np.exp(-x1 ** 2)
plt.fill_between([x0,x1],[y0,y1])

plt.xlim([-0.5,1.5]); plt.ylim([0,1.5])
plt.show()
```

Aproksimirat ćemo integral površinom trapezoida:

```python
A = 0.5 * (y1 + y0) * (x1 - x0)
print("Površina trapezoida:", A)
```

## Definicija

Pravilo za $N$ podintervala intervala $[a,b]$ istih duljina je:

$$ T_{N}(f) = \frac{\Delta x}{2}\sum_{i=1}^{N}[f(x_{i})+f(x_{i-1})]$$

gdje je $\Delta x = (b - a)/N$ duljina podintervala i $x_{i} = a + i\Delta x$.

## Implementacija

Napisat ćemo funkciju imena `trapz` koja će kao ulazne parametre uzimati $f$, $a$, $b$ i $N$ te će vratiti aproksimaciju $T_{N}(f)$.

```python
def trapz(f,a,b,N=50):
    '''
    Approximate the integral of f(x) from a to b by the trapezoid rule.

    The trapezoid rule approximates the integral int_a^b f(x) dx by the sum:
    (dx/2) sum_{k=1}^N (f(x_k) + f(x_{k-1}))
    where x_k = a + k*dx and dx = (b - a)/N.

    Parameters
    ----------
    f : function
        Vectorized function of a single variable
    a , b : numbers
        Interval of integration [a,b]
    N : integer
        Number of subintervals of [a,b]

    Returns
    -------
    float
        Approximation of the integral of f(x) from a to b using the
        trapezoid rule with N subintervals of equal length.

    Examples
    --------
    >>> trapz(np.sin,0,np.pi/2,1000)
    0.9999997943832332
    '''
    x = np.linspace(a,b,N+1) # N+1 points make N subintervals
    y = f(x)
    y_right = y[1:] # right endpoints
    y_left = y[:-1] # left endpoints
    dx = (b - a)/N
    T = (dx/2) * np.sum(y_right + y_left)
    return T
```

Testirajmo funkciju na poznatom integralu

$$\int_{0}^{\pi/2}\sin xdx = 1$$

```python
trapz(np.sin,0,np.pi/2,1000)
```

Testirajmo funkciju na još jednom poznatom integralu

$$\int_{0}^{1}3x^{2}dx = 1$$

```python
trapz(lambda x : 3*x**2,0,1,10000)
```

I na još jednom poznatom integralu:

$$\int_{0}^{1}x dx = \frac{1}{2}$$

```python
trapz(lambda x : x,0,1,1)
```

## `scipy.integrate.trapezoid`

U paketu [SciPy](https://docs.scipy.org/doc/scipy/index.html) postoji funkcija [scipy.integrate.trapezoid](https://docs.scipy.org/doc/scipy/reference/generated/scipy.integrate.trapezoid.html) koja služi za aproksimaciju integrala korištenjem trapezoida.

Pokušajmo aproksimirati funkciju $f(x) = \frac{1}{1 + x^{2}}$ na intervalu $[0, 5]$ uz $N = 10$.

```python
f = lambda x : 1/(1 + x**2)
a = 0; b = 5; N = 100

# x and y values for the trapezoid rule
x = np.linspace(a,b,N+1)
y = f(x)

# X and Y values for plotting y=f(x)
X = np.linspace(a,b,100)
Y = f(X)
plt.plot(X,Y)

for i in range(N):
    xs = [x[i],x[i],x[i+1],x[i+1]]
    ys = [0,f(x[i]),f(x[i+1]),0]
    plt.fill(xs,ys,'b',edgecolor='b',alpha=0.2)

plt.title('Trapezoid Rule, N = {}'.format(N))
plt.show()
```

Sad ćemo izračunati aproksimaciju ukupne površine ispod krivulje korištenjem funkcije `trapezoid`

```python
from scipy import integrate

T = integrate.trapezoid(Y, X)
print(T)
```

Za taj integral nam je poznata točna vrijednost

$$\int_{0}^{5}\frac{1}{1 + x^{2}}dx = \arctan(5)$$

Pa možemo usporediti vrijednosti:

```python
I = np.arctan(5)
print(I)
```

```python
print(f"Razlika između točne i procijenjene vrijednosti integrala: {np.abs(I - T)}")
```
