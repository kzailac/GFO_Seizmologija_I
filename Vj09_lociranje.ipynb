Za trijangulaciju je potrebno nacrtati kružnicu. Radi jednostavnosti, želimo koristiti Kartezijev koordinatni sustav, ali koordinate postaja su u WGS sustavu!

Na području RH je uobičajeno koristiti [HTRS sustav](https://epsg.io/3765). Za konverziju između koordinatnih sustava koristit ćemo paket [`pyproj`](https://pypi.org/project/pyproj/).

```python
from pyproj import Transformer

# inicijaliziramo koordinatne sustave
crs_wgs84 = 4326
crs_htrs = 3765

# definiramo funkciju koja ce raditi konverziju iz lat, lon u x, y
def wgs2htrs(lat, lon):
    transformer = Transformer.from_crs(crs_from=crs_wgs84, crs_to=crs_htrs)
    x, y = transformer.transform(lat, lon)
    return x / 1000, y / 1000
```

```python
x_ston, y_ston = wgs2htrs(lat_ston, lon_ston)
x_hvar, y_hvar = wgs2htrs(lat_hvar, lon_hvar)
x_ptj, y_ptj = wgs2htrs(lat_ptj, lon_ptj)
```

Nacrtajmo kružnice na kartu!

```python
import pandas as pd
import pygmt

stations = pd.DataFrame({
    "lon": [lon_ston, lon_hvar, lon_ptj],
    "lat": [lat_ston, lat_hvar, lat_ptj],
    "D": [d_ston, d_hvar, d_ptj]
})

ellipse = pd.DataFrame({
    "lon": stations["lon"],
    "lat": stations["lat"],
    "diameter_km": [2 * dd for dd in stations["D"]]
})

region = [12, 20, 41, 48.5]

fig = pygmt.Figure()

fig.coast(
    region=region,
    shorelines=True,
    land="white",
    water="lightblue",
    projection="M15c",
    frame="a",
    borders="a"
)

fig.plot(
    data=ellipse,
    style="E-",
    pen="1p,black"
)

fig.plot(
    x=stations["lon"],
    y=stations["lat"],
    style="t0.5c",
    fill="blue",
    pen="black",
    label="Postaje"
)

fig.show()
```

Jednadžba kružnice u Kartezijevom koordinatnom sustavu glasi:

$$(x - x_{0}) ^ {2} + (y - y_{0})^{2} = r^{2},$$

gdje su $x_{0},\;y_{0}$ koordinate središta kružnice, a $r$ je njezin polumjer.

Pronađimo sjecišta kružnica. Koristit ćemo funkciju `circle_intersections` (jednadžbe izvedene na satu!).

```python
import math

def circle_intersections(x1, y1, r1, x2, y2, r2):
    a0 = (r2 ** 2 - r1 ** 2 + x1 ** 2 - x2 ** 2 + y1 ** 2 - y2 ** 2) / (2 * (y1 - y2))
    b0 = - (x1 - x2) / (y1 - y2)

    a = 1 + b0 ** 2
    b = 2 * (a0 - y1) * b0 - 2 * x1
    c = x1 ** 2 + (a0 - y1) ** 2 - r1 ** 2

    D = b ** 2 - 4 * a * c
    x1 = (-b + math.sqrt(D)) / (2 * a)
    x2 = (-b - math.sqrt(D)) / (2 * a)

    y1 = a0 + b0 * x1
    y2 = a0 + b0 * x2

    P1 = (x1, y1)
    P2 = (x2, y2)

    return P1, P2
```

```python
P1_ston_hvar, P2_ston_hvar = circle_intersections(
    x_ston, y_ston, d_ston, x_hvar, y_hvar, d_hvar
) 
P1_hvar_ptj, P2_hvar_ptj = circle_intersections(
    x_hvar, y_hvar, d_hvar, x_ptj, y_ptj, d_ptj
)
P1_ston_ptj, P2_ston_ptj = circle_intersections(
    x_ston, y_ston, d_ston, x_ptj, y_ptj, d_ptj
)

# za prikaz na slici su potrebne koordinate u WGS sustavu
def htrs2wgs(x, y):
    transformer = Transformer.from_crs(crs_from=crs_htrs, crs_to=crs_wgs84)
    lat, lon = transformer.transform(x * 1000, y * 1000)
    return lat, lon

p11_lat, p11_lon = htrs2wgs(P1_ston_hvar[0], P1_ston_hvar[1])
p12_lat, p12_lon = htrs2wgs(P2_ston_hvar[0], P2_ston_hvar[1])
p21_lat, p21_lon = htrs2wgs(P1_hvar_ptj[0], P1_hvar_ptj[1])
p22_lat, p22_lon = htrs2wgs(P2_hvar_ptj[0], P2_hvar_ptj[1])
p31_lat, p31_lon = htrs2wgs(P1_ston_ptj[0], P1_ston_ptj[1])
p32_lat, p32_lon = htrs2wgs(P2_ston_ptj[0], P2_ston_ptj[1])
```

```python
fig.plot(
    x=[
        p11_lon,
        p12_lon,
        p21_lon,
        p22_lon,
        p31_lon,
        p32_lon
    ],
    y=[
        p11_lat,
        p12_lat,
        p21_lat,
        p22_lat,
        p31_lat,
        p32_lat
    ],
    style="c0.25c",
    fill="black",
    pen="black"
)

fig.text(
    x=[
        p11_lon,
        p12_lon,
        p21_lon,
        p22_lon,
        p31_lon,
        p32_lon
    ],
    y=[
        p11_lat,
        p12_lat,
        p21_lat,
        p22_lat,
        p31_lat,
        p32_lat
    ],
    text=["P11", "P12", "P21", "P22", "P31", "P32"],
    justify="BL"
)
fig.show()
```

Izračunajmo sada koordinate epicentra potresa kao srednjak triju sjecišta najbližih sredini. Za račun koristimo Kartezijeve koordinate, za crtanje koristimo WGS koordinate.

```python
x_e = (P1_ston_hvar[0] + P1_hvar_ptj[0] + P2_ston_ptj[0]) / 3
y_e = (P1_ston_hvar[1] + P1_hvar_ptj[1] + P2_ston_ptj[1]) / 3

lat_e, lon_e = htrs2wgs(x_e, y_e)
```

Konačno nacrtamo i epicentar na kartu!

```python
fig.plot(
    x=lon_e,
    y=lat_e,
    style="a0.5c",
    fill="red",
    pen="black"
)
fig.show()
```

## 5. korak

Sad kad imamo definirane pickove, možemo ih nacrtati na seizmograme! 

```python
def plot_picks(st_orig, picks):
    st_sta = st_orig.copy()
    starttime = st_sta[0].stats.starttime

    p_pick = starttime + picks["p_pick"]
    s_pick = starttime + picks["s_pick"]
    
    st_sta.trim(starttime=p_pick - 10, endtime=s_pick + 40)

    p_pick = p_pick.datetime
    s_pick = s_pick.datetime

    z_comp = st_sta.select(channel="BHZ")[0]
    n_comp = st_sta.select(channel="BHN")[0]
    e_comp = st_sta.select(channel="BHE")[0]
    
    fig, axs = plt.subplots(3, 1, sharex=True, figsize=(10, 8))
    
    axs[0].plot(z_comp.times(type="matplotlib"), z_comp.data, color="black")
    axs[0].plot(2 * [p_pick], [z_comp.data.min(), z_comp.data.max()], color="red")
    axs[0].plot(2 * [s_pick], [z_comp.data.min(), z_comp.data.max()], color="blue")
    axs[0].xaxis_date()  
    
    axs[1].plot(n_comp.times(type="matplotlib"), n_comp.data, color="black")
    axs[1].plot(2 * [p_pick], [n_comp.data.min(), n_comp.data.max()], color="red")
    axs[1].plot(2 * [s_pick], [n_comp.data.min(), n_comp.data.max()], color="blue")
    axs[1].xaxis_date()
    
    axs[2].plot(e_comp.times(type="matplotlib"), e_comp.data, color="black")
    axs[2].plot(2 * [p_pick], [e_comp.data.min(), e_comp.data.max()], color="red")
    axs[2].plot(2 * [s_pick], [e_comp.data.min(), e_comp.data.max()], color="blue")
    axs[2].xaxis_date()
    
    fig.autofmt_xdate()
    
    plt.show();

plot_picks(st_kaln, st_picks["KALN"])
plot_picks(st_mosl, st_picks["MOSL"])
plot_picks(st_ston, st_picks["STON"])
```
